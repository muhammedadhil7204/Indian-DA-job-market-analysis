```python
import pandas as pd

df = pd.read_csv('NaukriData_data analytics.csv')
print(df.shape)
print(df.columns.tolist())
print(df.head())
```

```python
skills_list = [
    'SQL', 'Python', 'SAS', 'Excel', 'Power BI', 'Tableau',
    'Data Analysis', 'Data Visualization', 'Machine Learning',
    'Data cleansing', 'Data management', 'Data quality', 'Data analytics',
    'Risk management', 'Risk analysis', 'Statistics', 'VBA'
]

skill_counts = {}
for skill in skills_list:
    count = df['Skills'].str.contains(skill, case=False, na=False).sum()
    skill_counts[skill] = count

for skill, count in sorted(skill_counts.items(), key=lambda x: -x[1]):
    print(f"{skill}: {count}")
    
```
